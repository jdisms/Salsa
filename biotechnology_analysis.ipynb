# Salsa Biotechnology Analysis Example

This notebook demonstrates basic biotechnology analysis using Python. It's a great starting point for your biotechnology projects!

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from dna_analyzer import DNAAnalyzer

# Set up plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

print("🧬 Salsa Biotechnology Analysis Environment Ready!")
```

## 1. DNA Sequence Analysis

Let's analyze some DNA sequences using our custom DNA analyzer.

```python
# Initialize DNA analyzer
analyzer = DNAAnalyzer()

# Sample DNA sequences
sequences = {
    "Sample_1": "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGTAG",
    "Sample_2": "ATGAAATTTGCAGCAATCTGCAAGAAATGCAAGCAATCTGCAAGAAATGCAAGCAATAG",
    "Sample_3": "ATGGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGCCTAG",
    "Sample_4": "ATGAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAAATAG",
    "Sample_5": "ATGCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCTAG"
}

# Analyze each sequence
results = []
for name, seq in sequences.items():
    analysis = analyzer.analyze_composition(seq)
    analysis['sample'] = name
    results.append(analysis)

# Convert to DataFrame
df = pd.DataFrame(results)
print("Sequence Analysis Results:")
print(df[['sample', 'length', 'gc_content', 'at_content']])
```

## 2. Data Visualization

Let's create some visualizations of our sequence data.

```python
# Create subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('DNA Sequence Analysis Results', fontsize=16, y=0.98)

# Plot 1: GC Content
axes[0, 0].bar(df['sample'], df['gc_content'], color='skyblue', alpha=0.7)
axes[0, 0].set_title('GC Content by Sample')
axes[0, 0].set_ylabel('GC Content (%)')
axes[0, 0].tick_params(axis='x', rotation=45)

# Plot 2: AT Content
axes[0, 1].bar(df['sample'], df['at_content'], color='lightcoral', alpha=0.7)
axes[0, 1].set_title('AT Content by Sample')
axes[0, 1].set_ylabel('AT Content (%)')
axes[0, 1].tick_params(axis='x', rotation=45)

# Plot 3: Sequence Length
axes[1, 0].bar(df['sample'], df['length'], color='lightgreen', alpha=0.7)
axes[1, 0].set_title('Sequence Length by Sample')
axes[1, 0].set_ylabel('Length (bp)')
axes[1, 0].tick_params(axis='x', rotation=45)

# Plot 4: GC vs AT Content Scatter
axes[1, 1].scatter(df['gc_content'], df['at_content'], 
                   s=100, alpha=0.7, c=range(len(df)), cmap='viridis')
axes[1, 1].set_title('GC vs AT Content')
axes[1, 1].set_xlabel('GC Content (%)')
axes[1, 1].set_ylabel('AT Content (%)')

# Add sample labels to scatter plot
for i, row in df.iterrows():
    axes[1, 1].annotate(row['sample'], 
                       (row['gc_content'], row['at_content']),
                       xytext=(5, 5), textcoords='offset points', fontsize=8)

plt.tight_layout()
plt.show()
```

## 3. Base Composition Analysis

Let's create a detailed composition analysis for each sequence.

```python
# Create composition data
composition_data = []
for result in results:
    sample = result['sample']
    for base, count in result['composition'].items():
        composition_data.append({
            'sample': sample,
            'base': base,
            'count': count,
            'percentage': (count / result['length']) * 100
        })

comp_df = pd.DataFrame(composition_data)

# Create stacked bar chart
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Stacked bar chart - counts
pivot_counts = comp_df.pivot(index='sample', columns='base', values='count')
pivot_counts.plot(kind='bar', stacked=True, ax=ax1, 
                 color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'])
ax1.set_title('Base Composition (Counts)')
ax1.set_ylabel('Count')
ax1.tick_params(axis='x', rotation=45)
ax1.legend(title='Base')

# Stacked bar chart - percentages
pivot_percent = comp_df.pivot(index='sample', columns='base', values='percentage')
pivot_percent.plot(kind='bar', stacked=True, ax=ax2,
                  color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'])
ax2.set_title('Base Composition (Percentages)')
ax2.set_ylabel('Percentage (%)')
ax2.tick_params(axis='x', rotation=45)
ax2.legend(title='Base')

plt.tight_layout()
plt.show()

# Display composition table
print("\nDetailed Base Composition:")
print(pivot_percent.round(2))
```

## 4. Protein Translation Analysis

Let's translate our DNA sequences to proteins and analyze the results.

```python
# Translate sequences to proteins
protein_data = []
for name, seq in sequences.items():
    for frame in [1, 2, 3]:
        protein = analyzer.translate_to_protein(seq, frame)
        protein_data.append({
            'sample': name,
            'frame': frame,
            'protein': protein,
            'length': len(protein),
            'has_stop': '*' in protein
        })

protein_df = pd.DataFrame(protein_data)

# Display protein translation results
print("Protein Translation Results:")
print("============================")
for _, row in protein_df.iterrows():
    print(f"Sample: {row['sample']}, Frame: {row['frame']}")
    print(f"Protein: {row['protein'][:50]}{'...' if len(row['protein']) > 50 else ''}")
    print(f"Length: {row['length']} amino acids, Has stop codon: {row['has_stop']}")
    print("-" * 60)

# Visualize protein lengths
fig, ax = plt.subplots(figsize=(12, 6))
sns.barplot(data=protein_df, x='sample', y='length', hue='frame', ax=ax)
ax.set_title('Protein Length by Sample and Reading Frame')
ax.set_ylabel('Protein Length (amino acids)')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

## 5. Statistical Analysis

Let's perform some basic statistical analysis on our sequence data.

```python
# Calculate basic statistics
stats_summary = df[['length', 'gc_content', 'at_content']].describe()
print("Descriptive Statistics:")
print(stats_summary.round(2))

# Calculate correlation matrix
correlation_matrix = df[['length', 'gc_content', 'at_content']].corr()
print("\nCorrelation Matrix:")
print(correlation_matrix.round(3))

# Create correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
            square=True, linewidths=0.5)
plt.title('Correlation Matrix of Sequence Properties')
plt.tight_layout()
plt.show()
```
